```python
import os
from google.colab import userdata

os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.messages import HumanMessage, SystemMessage, AIMessage
from langchain.tools import tool
from langchain.agents import create_agent
```

```python
import requests

@tool
def get_weather(location: str):
    """ This is a function for geting the live weather of any loaction"""
    try:
        url = "https://wttr.in/" + location + "?format=j2"
        return requests.get(url).text
    except requests.exceptions.HTTPError as err:
        print(err)
```

```python
@tool
def get_ideas():
  """This function gives idea on the story or poem if nessasory"""
  return "Thriller, Comedy, Suspence"
```

```python
agent = create_agent(model = "google_genai:gemini-3.5-flash-lite",
                     system_prompt = SystemMessage(content="Behave like author and write the beautiful and molodious poems and stories"),
                     tools = [get_weather, get_ideas],
                     middleware=[])
agent
```

```python
response = agent.invoke({"messages":[HumanMessage(content="i need a story on the andhra")]})
```

```python
response["messages"][-1].text
```

```python
from gtts import gTTS
```

```python
tts = gTTS(response["messages"][-1].text, lang ="te")
tts.save("weather.mp3")
```
