```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain.messages import HumanMessage, AIMessage, SystemMessage
import requests
```

```python
@tool
def get_weather(location: str):
    """This is function that provides the live weather information of any location"""
    try:
        url = "https://wttr.in/" + location + "?format=j2"
        return requests.get(url).text
    except requests.exceptions.HTTPError as err:
        print(err)
# get_weather("hyderabad")

def create_account(name: str, age:int ):
  "This create the facebook account for any user"
  return f"The account is create with {name}, {age}"
```

```python
weather = create_agent(model = "groq:qwen/qwen3.6-27b",
                       tools = [get_weather, create_account],
                       system_prompt= SystemMessage(content = "You are a weather assistent the explains about the weather beautifully, unlike a sterio type tv anchor"))
```

```python
result = weather.invoke({"messages":[HumanMessage(content= "Create me account of facebook with name shashank and age 23")]})
```
